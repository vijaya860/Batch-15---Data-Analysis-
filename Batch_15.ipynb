```python
# Prepare Data for Basic Sales Forecasting

from sklearn.linear_model import LinearRegression
import numpy as np

forecast_data = monthly_sales.copy()

# Create sequential time index
forecast_data["Time"] = np.arange(len(forecast_data))

print(forecast_data.head())
```

```python
# Basic Sales Forecasting using Linear Regression

X = forecast_data[["Time"]]
y = forecast_data["Sales"]

model = LinearRegression()
model.fit(X, y)

print("Linear Regression model trained successfully.")
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Forecast next 6 months

future_time = pd.DataFrame({
    "Time": np.arange(
        len(forecast_data),
        len(forecast_data) + 6
    )
})

future_sales = model.predict(future_time)

print("Forecasted Sales for Next 6 Months:")

for i, sales in enumerate(future_sales, start=1):
    print(f"Month {i}: {sales:.2f}")
```

```python
# Actual vs Forecasted Sales

future_dates = pd.date_range(
    start=forecast_data["Date"].iloc[-1] + pd.DateOffset(months=1),
    periods=6,
    freq="MS"
)

plt.figure(figsize=(12, 5))

# Actual sales
plt.plot(
    forecast_data["Date"],
    forecast_data["Sales"],
    marker="o",
    label="Actual Sales"
)

# Forecasted sales
plt.plot(
    future_dates,
    future_sales,
    marker="o",
    linestyle="--",
    label="Forecasted Sales"
)

plt.title("Actual vs Forecasted Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
